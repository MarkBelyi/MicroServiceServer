### Imports

```python
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

### Data

days_to_expiry (количество дней до истечения срока годности),
stock_left (остаток товара на складе),
predicted_demand (прогнозируемый спрос).

Целевая переменная (target) будет бинарной: 1 — продукт будет продан до истечения срока годности, 0 — не будет.

```python
## В будущем перейти на адекватные данные, а пока закинем синтетику

np.random.seed(13)

#Данные
days_to_expiry = np.random.randint(1, 365, size=1000)  # Количество дней до истечения срока годности
stock_left = np.random.randint(1, 1000, size=1000)  # Остаток на складе
predicted_demand = np.random.randint(1, 500, size=1000)  # Прогнозируемый спрос
```

```python
target = (predicted_demand > stock_left / 2).astype(int)

# Нужна бинарная переменная

'''
Если прогнозируемый спрос (predicted_demand) = 300, а остаток товара на складе (stock_left) = 400:
Условие: 300 > 400 / 2, т.е. 300 > 200. Это True, и целевая переменная будет 1 (предполагается, что продукт будет продан).
Если прогнозируемый спрос = 100, а остаток товара = 400:
Условие: 100 > 200 — это False, и целевая переменная будет 0 (продукт не будет продан).
'''
```

```python
# Здесь создаем табличку
X = np.column_stack((days_to_expiry, stock_left, predicted_demand))
y = target
```

```python
# Делим
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=13)
```

```python
# Стандартизируем данные
## Formula: z = x - mean / std
'''
    Стандартизация помогает:
Ускорить сходимость градиентных методов.
Убедиться, что алгоритм учитывает все признаки равномерно.
Улучшить производительность и точность моделей машинного обучения.
'''
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Model

```python
# Определение модели
model = Sequential()

# Входной слой и первый скрытый слой
model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))

# Второй скрытый слой
model.add(Dense(32, activation='relu'))

# Выходной слой с активацией sigmoid для предсказания вероятности
model.add(Dense(1, activation='sigmoid'))

# Компиляция модели
model.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])
```

## Train

```python
# Обучение модели
history = model.fit(X_train_scaled, y_train, epochs=50, batch_size=32, validation_split=0.2)
```

## Grading

```python
# Оценка модели на тестовых данных
loss, accuracy = model.evaluate(X_test_scaled, y_test)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

## Predict

```python
# Пример предсказания для нового продукта
new_product_data = np.array([[30, 100, 50]])  # days_to_expiry=30, stock_left=100, predicted_demand=50
new_product_data_scaled = scaler.transform(new_product_data)
predicted_probability = model.predict(new_product_data_scaled)
print(f"Вероятность продажи продукта до истечения срока годности: {predicted_probability[0][0] * 100:.2f}%")
```
